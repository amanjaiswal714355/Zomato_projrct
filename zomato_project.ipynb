```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
#step 2
```

```python
dataframe = pd.read_csv("Zomato data .csv")
print(dataframe)
```

```python
dataframe
```

```python
#conver the data typr of column - rate
```

```python
def handleRate(value):
    value = str(value).split('/')
    value = value[0];
    return float(value)

dataframe['rate']=dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
dataframe.info()
```

```python
# type of resturant
```

```python
dataframe.head()
```

```python
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("type of resturant")
```

```python
#conclusion - majority of the resturant falls in dinning category
```

```python
dataframe.head()
```

```python
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes':grouped_data})
plt.plot(result, c="green", marker="o")
plt.xlabel("type of restaurant", c="red", size=20)
plt.ylabel("votes", c="red", size=20)
```

```python
#conclusion - dinning resturants has recieved maximum votes
```

```python
dataframe.head()
```

```python
plt.hist(dataframe['rate'],bins =5)
plt.title("ratings distribution")
plt.show()
```

```python
#conclusion - the majority resturants received ratings from 3.5 to 4
```

```python
couple_data=dataframe['approx_cost(for two people)']
sns.countplot(x=couple_data)
```

```python
#which mode receives maximum rating
```

```python
dataframe.head()
```

```python
plt.figure(figsize = (6,6))
sns.boxplot(x = 'online_order', y = 'rate', data =dataframe)
```

```python
#conclusion - offline order received lower rating in comparison to onlin order
```

```python
pivot_table = dataframe.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap="YlGnBu", fmt='d')
plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("Listed In (type)")
plt.show()
```

```python
#Conclusion: Dining resturants primarily accept offline orders,whereas cafes primarily receive online orders. This suggests that clirnts prefers orders in person at restaurants, but prefer online ordering at cafes.
```
